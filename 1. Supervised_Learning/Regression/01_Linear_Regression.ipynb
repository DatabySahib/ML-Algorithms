```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df['mileage'] = df['mileage'].str.split(' ').str[0]
df['mileage'] = pd.to_numeric(df['mileage'], errors='coerce')
```

```python
df['engine'] = df['engine'].str.split(' ').str[0]
df['engine'] = pd.to_numeric(df['engine'], errors='coerce')
```

```python
df['max_power'] = df['max_power'].str.split(' ').str[0]
df['max_power'] = pd.to_numeric(df['max_power'], errors='coerce')
```

```python
print(df[['mileage', 'engine', 'max_power']].dtypes)
```

```python
df.isnull().sum()
```

```python
cols = ['mileage', 'max_power', 'engine']
```

```python
for col in cols:
  df[col] = df[col].fillna(df[col].median())
```

```python
df['seats'] = df['seats'].fillna(df['seats'].mode()[0])
```

```python
df = df.drop('torque', axis=1)
```

```python
df.isna().sum()
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1,2,1)
sns.histplot(df['selling_price'], kde=True, color='blue')
plt.title('Maşın Qiymətlərinin Paylanması')
plt.xlabel('Satış Qiyməti')

plt.subplot(1,2,2)
sns.boxplot(x=df['selling_price'], color='red')
plt.title('Qiymət Sütunundakı Outlier-lər')
plt.xlabel('Satış Qiyməti')

plt.show()
```

```python
Q1 = df['selling_price'].quantile(0.25)
Q3 = df['selling_price'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print('Lower Limit:', lower_limit)
print('Upper Limit:', upper_limit)
```

```python
if lower_limit < 0:
  lower_limit = 0
```

```python
print('Lower Limit:', lower_limit)
print('Upper Limit:', upper_limit)
```

```python
df = df[(df['selling_price'] >= lower_limit) & (df['selling_price'] <= upper_limit)]
```

```python
df.shape
```

```python
numeric_cols = df.select_dtypes(include=['float64', 'int64'])

plt.figure(figsize=(10, 6))
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Rəqəmsal Sütunların Bir-biri ilə Korrelyasiyası')

plt.show()
```

```python
df['car_age'] = 2026 - df['year']
```

```python
df = df.drop(['name', 'year'], axis=1, errors='ignore')
```

```python
df = pd.get_dummies(df, drop_first=True)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
X = df.drop('selling_price', axis=1)
y = df['selling_price']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size = .2,
    random_state = 42
)
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print('MAE:', mean_absolute_error(y_test, y_pred))
print('MSE:', mean_squared_error(y_test, y_pred))
print('R2 Score:', r2_score(y_test, y_pred))
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(x=y_test, y=y_pred, alpha=0.5, color='purple')

plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color='red', lw=2, linestyle='--')

plt.xlabel('Həqiqi Qiymətlər')
plt.ylabel('Modelin Təxminləri (Predicted)')
plt.title('Həqiqi Qiymətlər vs Model Təxminləri')
plt.show()
```
